```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

f=files.upload()
d=pd.read_csv(next(iter(f)),header=None)

X=d[0].values
y=d[2].values

b=m=0
c=[]

for i in range(1500):
    e=b+m*X-y
    b-=.00000001*np.mean(e)
    m-=.00000001*np.mean(e*X)
    c.append(np.mean(e**2))

print("Intercept:",b)
print("Slope:",m)

plt.scatter(X,y)
plt.plot(X,b+m*X)
plt.show()

plt.plot(c)
plt.show()
```
