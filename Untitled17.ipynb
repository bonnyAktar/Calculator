```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("p2.txt", header=None,
                 names=["Size", "Bedrooms", "Price"])

X = df.iloc[:, :2].values
y = df.iloc[:, 2].values
m = len(y)

mean = X.mean(axis=0)
std = X.std(axis=0)
X = (X - mean) / std
X = np.c_[np.ones(m), X]

theta = np.zeros(3)
loss = []

for i in range(1000):
    error = X @ theta - y
    theta -= 0.01 * (X.T @ error) / m
    loss.append(np.mean(error**2))

print("Theta:", theta)

plt.plot(loss)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.show()

pred = X @ theta

plt.scatter(df["Size"], y)
plt.scatter(df["Size"], pred)
plt.xlabel("Size")
plt.ylabel("Price")
plt.show()

house = (np.array([1360, 3]) - mean) / std
house = np.r_[1, house]

print("Predicted Price:", house @ theta)
```
