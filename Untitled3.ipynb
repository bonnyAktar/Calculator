```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.datasets import make_blobs

# Generate data
X, _ = make_blobs(n_samples=20, centers=4, cluster_std=1.0, random_state=42)

# Display data points
plt.figure(figsize=(10, 10))
plt.scatter(X[:, 0], X[:, 1], s=100)

for i, p in enumerate(X):
    plt.annotate(f"P{i+1}", p, xytext=(5, 5), textcoords="offset points")

plt.title("Sample Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

# Dendrograms
methods = ["single", "complete", "average"]

plt.figure(figsize=(18, 5))

for i, method in enumerate(methods):
    plt.subplot(1, 3, i + 1)
    dendrogram(
        linkage(X, method=method),
        labels=[f"P{i+1}" for i in range(len(X))],
        leaf_rotation=90
    )
    plt.title(method.title() + " Linkage")
    plt.xlabel("Data Points")
    plt.ylabel("Distance")

plt.tight_layout()
plt.show()
```
